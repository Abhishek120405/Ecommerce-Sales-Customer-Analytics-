# E-Commerce Sales & Customer Analytics
Synthetic practice dataset: 600 orders in 2025. Profit is estimated.

```
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd() if (Path.cwd() / 'data' / 'sales_data.csv').exists() else Path.cwd().parent
DATA = ROOT / 'data' / 'sales_data.csv'
df = pd.read_csv(DATA, parse_dates=['Order_Date'])
df.head()
```

## Data quality checks

```
print('Shape:', df.shape)
print('Duplicate rows:', df.duplicated().sum())
display(df.isna().sum().to_frame('missing_values'))
```

## KPI summary

```
delivered = df[df['Order_Status'].eq('Delivered')]
kpis = {'Total Orders': df.Order_ID.nunique(), 'Delivered Orders': delivered.Order_ID.nunique(), 'Unique Customers': df.Customer_ID.nunique(), 'Delivered Revenue INR': delivered.Revenue_INR.sum(), 'Estimated Delivered Profit INR': delivered.Profit_INR.sum(), 'Average Delivered Order Value INR': delivered.Revenue_INR.sum()/max(delivered.Order_ID.nunique(),1)}
pd.Series(kpis)
```

## Revenue breakdowns

```
for col in ['Region','Category','Sales_Channel','Product']:
    display(delivered.groupby(col,as_index=False).agg(Orders=('Order_ID','nunique'),Revenue_INR=('Revenue_INR','sum')).sort_values('Revenue_INR',ascending=False).head(10))
```

## Monthly trend

```
monthly = delivered.assign(Month=delivered.Order_Date.dt.to_period('M').astype(str)).groupby('Month',as_index=False).Revenue_INR.sum()
ax=monthly.plot(x='Month',y='Revenue_INR',kind='line',marker='o',legend=False,title='Monthly Delivered Revenue')
ax.set_ylabel('Revenue (INR)'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Interpretation
Use the generated tables to identify the strongest regions, products, categories and channels. These findings apply only to this synthetic dataset.
